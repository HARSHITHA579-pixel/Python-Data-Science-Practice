**Imbalanced Dataset**

A dataset with unequal class distribution

```
#importing the dependancies
import numpy as np
import pandas as pd
```

```
#loading the dataset to pandas data frame
credit_card_dataset = pd.read_csv('/content/credit_data.csv')
credit_card_dataset.head()
```

```
credit_card_dataset.tail()
```

```
#determine distribution of two transactions
credit_card_dataset['Class'].value_counts()
```

**This is highly imbalanced data**

0 --> Legit Transactions
1 --> Fradulent transactions

```
#seperating the legit and fradulent transctions
legit = credit_card_dataset[credit_card_dataset.Class == 0]
fraud = credit_card_dataset[credit_card_dataset.Class == 1]
print(legit.shape)
print(fraud.shape)
```

UNDER sampling

Build a sample dataset containing similar distribution of Legit and fradulent transactions

Fradulent Transactions --> 224

```
legit_sample = legit.sample(n=224)
```

```
print(legit_sample.shape)
```

```
new_dataset = pd.concat([legit_sample, fraud], axis=0)
new_dataset.head()
```
